# Utils

## List checkpoint paths

```python
import glob
import os

approach = 'resized'

dataset_name = 'structseg_thorax'

base_ckpt_paths = os.path.join(f'model_checkpoints/{dataset_name}', '**', f'ckpt_official_sibgrapi_{approach}_*')

all_ckpts = sorted(glob.glob(base_ckpt_paths, recursive=True))

for c in all_ckpts:
    print(c)
```

## List preds paths

```python
import glob
import os
import pandas as pd

approach = 'resized'
dataset_name = 'structseg_thorax'

base_pred_paths = os.path.join(f'predictions/{dataset_name}', '**', f'pred_official_sibgrapi_{approach}_*')
all_preds = sorted(glob.glob(base_pred_paths, recursive=True))

for p in all_preds:
    print(p)
```

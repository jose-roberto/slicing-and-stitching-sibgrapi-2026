```python
import torch

from vnet import VNet
from resunet50 import ResUNet50
from highresnet import HighRes3DNet
from skipdensenet import SkipDenseNet3D
from resnetmed import ResNetMed3D
from unetr import UNETR
```

```python
target = 'SkipDenseNet3D'

if target == 'VNet':
    model = VNet(in_channels=1, num_classes=2)
elif target == 'ResUNet50':
    model = ResUNet50(in_channels=1, num_classes=2)
elif target == 'HighRes3DNet':
    model = HighRes3DNet(
            1,
            2,
            initial_out_channels_power=4,
            layers_per_residual_block=2,
            residual_blocks_per_dilation=2,
            dilations=2,
            batch_norm=False,
            instance_norm=True,
            residual=True,
            padding_mode='constant',
            add_dropout_layer=True
    )
elif target == 'SkipDenseNet3D':
    model = SkipDenseNet3D(
            in_channels=1, 
            num_classes=2,
            growth_rate=16,
            block_config=(4, 4, 4, 4),
            num_init_features=32,
            drop_rate=0.1,
            bn_size=4
    )
elif target == 'ResNetMed3D':
    model = ResNetMed3D(
            in_channels=1,                     
            num_classes=2,                     
            layers=[3, 4, 6, 3],              
            block_inplanes=[64, 128, 256, 512],
            no_max_pool=False,
            shortcut_type='B'
    )
elif target == 'UNETR':
    model = UNETR(
            in_channels=1,
            out_channels=2,
            img_size=(128, 128, 128),
            patch_size=(16, 16, 16),
            embedding_dim=768,
            depth=12,
            n_heads=12,
            mlp_ratio=4.,
            qkv_bias=False,
            p_dropout=0.,
            attn_dropout=0.
    )

total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"Total de parâmetros: {total_params:,}")
print(f"Parâmetros treináveis: {trainable_params:,}")

try:

    dummy_input = torch.randn(1, 1, 128, 128, 128) 
    print(f"Tamanho da entrada gerada: {dummy_input.shape}")
    
    output = model(dummy_input)
    
    print(f"Tamanho da saída da rede: {output.shape}")
    
    if output.shape == (1, 2, 128, 128, 128):
        print("Resolução mantida.")
    else:
        print("Saída com tamanhos diferentes.")
        
except Exception as e:
    print(f"Erro: {e}")
```
